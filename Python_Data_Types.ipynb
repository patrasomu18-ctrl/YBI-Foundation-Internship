# Python Data Types

## **1. Numeric Types:**

- Integers (int): Whole numbers, positive or negative, without decimals.
- Floating-point numbers (float): Numbers with a decimal point.
- Complex numbers (complex):  Numbers with a real and imaginary part (e.g., 3 + 5j).

```python
# Integer
age = 25
runs = 150
type(age), type(runs)
```

```python
# Float
price = 99.99
temperature = 36.5
type(price), type(temperature)
```

```python
# Comple number
z = 2 + 3j
type(z)
```

## **2. Text Type:**
- Strings (str): Sequences of characters enclosed in single, double, or triple quotes.

```python
# String
name = "Aisha"
city = 'Varanasi'
type(name), type(city)
```

```python
# Multiline string
quote = """This is a
multiline string."""
quote
```

## **3. Sequence Types:**
- Lists (list): Ordered, mutable collections of items (can be of different data types).
- Tuples (tuple): Ordered, immutable collections of items.

```python
# List
colors = ["red", "green", "blue"]
expenses = [500, 1200, 350.75]
type(colors), type(expenses)
```

```python
# Tuple
coordinates = (10, 20)
days_of_week = ("Monday", "Tuesday", "Wednesday")
type(coordinates), type(days_of_week)
```

## **4. Mapping Type:**

- Dictionaries (dict): Unordered collections of key-value pairs.
Python

```python
# Dictionary
student = {"name": "Riya", "age": 18, "city": "Delhi"}
type(student)
```

## **5. Set Types:**

- Sets (set): Unordered collections of unique items.
- Frozen sets (frozenset): Immutable sets.

```python
# Set
unique_numbers = {1, 2, 3, 4, 4, 5}
unique_numbers
```

```python
# Frozenset
numbers = frozenset((1,2,3))
numbers
```

## **6. Boolean Type:**
- Booleans (bool): Represents truth values (True or False).

```python
# Boolean
is_raining = True
is_adult = False
is_raining, is_adult
```

## **7. Binary Types:**

- Bytes (bytes), Byte Arrays (bytearray), Memory Views (memoryview): Used to handle binary data.

```python
# Bytes
byte_data = b"Hello"
print(byte_data)
```

## **8. None Type:**
- None: Represents the absence of a value.

```python
type(None)
```

```python
# 1. Numeric Types
age = 25              # Integer (int)
price = 99.99         # Float (float)
z = 3 + 5j            # Complex number (complex)

# 2. Sequence Types
course_name = "Python"                       # String (str)
fruits = ["apple", "banana", "mango"]        # List (list)
coordinates = (10.0, 20.0)                   # Tuple (tuple)

# 3. Mapping Type
student = {"name": "Soumyashree", "age": 19, "grade": "A"}  # Dictionary (dict)

# 4. Set Types
unique_numbers = {1, 2, 3, 4, 5}                     # Set (set)
immutable_set = frozenset(["apple", "banana", "mango"]) # Frozen Set (frozenset)

# 5. Boolean Type
is_student = True     # Boolean (bool)

print("Numeric:", age, price, z)
print("Sequences:", course_name, fruits, coordinates)
print("Dictionary:", student)
print("Sets:", unique_numbers, immutable_set)
print("Boolean:", is_student)
```

```python
# Define an integer and a float
age = 30
height = 5.9

# Perform arithmetic operations
new_age = age + 5
new_height = height * 2

print("New Age:", new_age)
print("New Height:", new_height)
```

```python
# Define a string and a list
name = "Priya"
friends = ["Adarsh", "Somu", "Disha"]

# Concatenate strings and access list elements
greeting = "Hello, " + name
first_friend = friends[0]

print(greeting)
print("First Friend:", first_friend)
```

```python
# Define a dictionary and a set
student = {"name": "Ararsh", "age": 22, "grade": "B"}
unique_skills = {"Python", "Java", "C++"}

# Access dictionary values and add to a set
student_name = student["name"]
unique_skills.add("JavaScript")

print("Student Name:", student_name)
print("Skills:", unique_skills)
```
