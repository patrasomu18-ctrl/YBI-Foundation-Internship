```python
# Python automatically infers the data type based on the value assigned
my_variable = 10                 # Integer
another_variable = "Hello, AI!"  # String

print("my_variable:", my_variable)
print("another_variable:", another_variable)
```

```python
# Descriptive variable names (snake_case and camelCase)
age = 21
total_price = 249.50
user_name = "Adarsh"

# Variable names are case-sensitive
myVariable = "Camel Case Value"
myvariable = "Lower Case Value"

print("Age:", age)
print("Total Price:", total_price)
print("User Name:", user_name)
print("myVariable:", myVariable)
print("myvariable:", myvariable)
```

```python
# This is a single-line comment

my_variable = 10  # This is an inline comment

"""
This is a multi-line comment.
It can span multiple lines.
"""

print("Variable value after comments:", my_variable)
```

```python
import keyword

# Print all reserved keywords that cannot be used as variable names
print("Reserved keywords in Python:")
print(keyword.kwlist)
```
