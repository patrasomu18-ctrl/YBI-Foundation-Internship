# Python Loops

```python
# 1. The 'for' Loop
ai_models = ['Neural Network', 'Decision Tree', 'Support Vector Machine']
for model in ai_models:
    print(f'Rohan is working on {model}')

# 2. The 'while' Loop
data_cleaned = False
iterations = 0
while not data_cleaned:
    print(f'Cleaning data... Iteration {iterations}')
    iterations += 1
    if iterations >= 5:
        data_cleaned = True

# 3. Nested Loops
algorithms = ['KNN', 'Random Forest']
datasets = ['Dataset A', 'Dataset B']
for algorithm in algorithms:
    for dataset in datasets:
        print(f'Evaluating {algorithm} on {dataset}')

# 4. Loop Control Statements (break and continue)
numbers = [1, 2, 3, 4, 5]
for number in numbers:
    if number == 3:
        continue
    if number == 5:
        break
    print(number)

# 5. Applications of Loops in AI
epochs = 10
for epoch in range(epochs):
    print(f'Training model: Epoch {epoch + 1}/{epochs}')
```
